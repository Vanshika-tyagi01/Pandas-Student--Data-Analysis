```python
df['Result']=df['Marks'].apply(lambda x:'Fail' if x <=40 else'Pass'if x <=60 else 'Good' if  x <=80 else'Very Good' if x <90 else 'Excellent')
print(df)
```

```python
df['Rank'] = df['Marks'].rank(ascending=False,method='min').astype(int)
```

```python
df['Grade']=df['Marks'].apply(lambda x : 'A+' if x >=90 else 'A' if x >=80 else 'B' if x >= 70 else 'c' if x >=60 else 'D')
```
